```python
variables_numericas = df2.select_dtypes(include=np.number).columns

Q1 = df2[variables_numericas].quantile(0.25)
Q3 = df2[variables_numericas].quantile(0.75)

IQR = Q3 - Q1

limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

outliers = (
    (df2[variables_numericas] < limite_inferior) |
    (df2[variables_numericas] > limite_superior)
).sum()

outliers
```
